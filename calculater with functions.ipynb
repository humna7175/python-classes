```python
def sum(a,b):
    return a+b
def subtract(a,b):
    return a-b
def multiply(a,b):
    return a*b
def division(a,b):
    return a/b

n1=int(input("enter number 1:"))
n2=int(input("enter number 2:"))

print("print the operation you want?")
print("1-add")
print("2-subtract")
print("3-multiply")
print("4-division")
op=int(input("enter the operater you want to perform"))

if op ==1:
    ans=sum(n1,n2)
elif op==2:
    ans=subtract(n1,n2)
elif op==3:
    ans=multiply(n1,n2)
elif op==4:
    ans=division(n1,n2)
else:
    print("please enter a valid operater")

print("the answe is:",ans)
    
     



    
```
