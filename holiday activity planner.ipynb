```python
# Holiday Activity Planner

holiday = input("Enter holiday type (Summer/Winter): ")

if holiday == "Summer":
    activity = input("Do you like Beach or Mountains? ")

    if activity == "Beach":
        print("Your holiday plan: Pack your swimsuit and enjoy a relaxing day at the beach!")
    else:
        print("Your holiday plan: Pack warm clothes and go hiking in the mountains!")

elif holiday == "Winter":
    activity = input("Do you like Snow or Indoors? ")

    if activity == "Snow":
        print("Your holiday plan: Enjoy skiing and build a snowman!")
    else:
        print("Your holiday plan: Stay indoors, read a book, and drink hot chocolate!")

else:
    print("Invalid holiday type. Please enter Summer or Winter.")
```
