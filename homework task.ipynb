```python
chores=[
    "complete history homework",
    "complete maths homework",
]
total_chores=len(chores)
current=0
completed=0
print("daily task check list")
while current<total_chores:
    print("task: ",chores[current])
    ans=input("have you completed your task?(yes/no")
    if ans=="yes":
        completed=completed+1
        current=current+1
        print("great,you completed you first chore")
    elif ans=="no":
        print("complete your task first,then try again")
    else:
        print("invaild")  
    print("completed:",completed)
```
