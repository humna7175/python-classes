```python
marks1=int(input("enter your english marks:"))
marks2=int(input("enter your geography marks:"))
marks3=int(input("enter your science marks:"))
total=marks1+marks2+marks3
avg=total/3
validrange=range(0,100)
if avg not in validrange:
    print("average is not valid")
elif avg in range(81,100):
    print("A grade")
elif avg in range(71,80):
    print("B grade")
elif avg in range(61,70):
    print("C grade")
else:
    print("fail")


```
