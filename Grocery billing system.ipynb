```python
customers = int(input("Enter customers: "))

c = 1

while c <= customers:

    print("Customer", c)

    items = int(input("Enter items: "))
    total = 0
    i = 1

    while i <= items:

        name = input("Enter item name: ")

        if name == "stop":
            break

        price = float(input("Enter price: "))

        if price <= 0:
            print("Invalid price")
            continue

        total = total + price
        i = i + 1

    print("Total Bill =", total)

    c = c + 1

    choice = input("Continue? yes/no: ")

    if choice == "n":
        break


print("Price Category Report")

categories = ["Low", "Medium", "High"]

for category in categories:

    print(category)

    for x in categories:

        if category == x:
            print("Checked:", x)
            

```
