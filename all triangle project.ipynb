```python

print("Pattern 1:")
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()

print("Pattern 2:")
for i in range(5):
    for j in range(5):
        print("#", end="")
    print()

print("Floyd's Triangle:")
number = 1

for i in range(1, 6):
    for j in range(i):
        print(number, end=" ")
        number += 1
    print()

print("Diamond Pattern:")

for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()

for i in range(4, 0, -1):
    for j in range(5 - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()
```
